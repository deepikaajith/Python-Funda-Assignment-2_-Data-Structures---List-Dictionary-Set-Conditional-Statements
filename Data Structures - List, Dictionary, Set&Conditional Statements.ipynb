```python
#a. Create a list named age_list with five integer elements. For eg., [24, 25, 26, 27, 28]
age_list = [24,25,26,27,28]

#b. Create a list named name_list with five string elements.
name_list = ["Maya" , "Meenu" , "Deva" , "Krishna" , "Anirudh"]

#2. List Operations / Modifications:
#a. Append the string "Yazhini" to name_list.
#using append Function I added the string "Yazhini"
name_list.append("Yazhini")
print(name_list)
#b. Insert the element 30 at index 2 in age_list.
#using insert function by giving the index position 2 , inserted the integer 30.
age_list.insert(2,30)
print(age_list)
#c. Remove the string "Yazhini" from name_list.
#using remove function , removed the string "Yazhini"
name_list.remove("Yazhini")
print(name_list)
#d. Pop the last element from age_list.
#using pop(), will remove the last element.
age_list.pop()
print(age_list)
#e. Extend the age_list with additional ages [29, 30, 26].
#using extend([]) function can add multiple string or values.
age_list.extend([29,30,26])
print(age_list)
#f. Sort age_list in descending order.
#For sorting in descending order I used reverse function inside sort and give True.
age_list.sort(reverse = True)
print(age_list)
#g. Find Max age, Min age and sum of all ages from age_list.
print(max(age_list))
print(min(age_list))
print(sum(age_list))
```

```python
#3. Accessing List Elements:
#a. Print the first element of name_list.
#gave the index number 0 in square bracket to print the first element.
print(name_list[0])
#b. Print the last element of name_list.
#gave the index number as -1 in square bracket to print the last element.
print(name_list[-1])
#c. Print the elements from index 2 to index 4 in name_list.
#Here I used slicing is used to get a range of elements from a list.
print(name_list[2:5])
#d. Print the elements of name_list in reverse order.
#Reverse function is used here
name_list.reverse()
print(name_list)
```

```python
#Dictionary (Creation, Modification and Access):
'''a.Create a dictionary named student_marks that maps the names of five
students to their marks (use scale of from 0 to 100).'''

Student_marks = {"Maya" : 35 ,"Rahul" : 55 ,"Deva": 20,
                 "Anju": 70 , "Kavya": 82}
print(Student_marks)

#b. Access and print the mark of a specific student, of your choice.
print(Student_marks["Maya"])

#c. Add a new student "Janani" with a mark of 80 to the student_marks dictionary.
#Here is use update function to add Janani with marks.
Student_marks.update ({"Janani" : 80})
print(Student_marks)

#d. Update the mark of any one older student to 82.
#Here I use alternate method of update to add new mark.
Student_marks["Rahul"] = 82
print(Student_marks)

'''e. Use the keys(), values(), and items() methods to print all keys, values, and
key-value pairs in the student_marks dictionary.'''
# Print all keys
print(Student_marks.keys())
# Print all values
print(Student_marks.values())
# Print all key-value pairs
print(Student_marks.items())

```
